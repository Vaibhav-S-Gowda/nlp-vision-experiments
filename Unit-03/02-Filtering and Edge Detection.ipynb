# Filtering and Edge Detection

```python
import cv2 as cv
import numpy as np

# Load the image 
image  = cv.imread(r'C:\Users\vaibh\Downloads\serene-samurai-amidst-cherry-blossoms-moonlit-lake.jpg', cv.IMREAD_GRAYSCALE)
cv2.imshow("Original Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
display_img = cv2.resize(img, (800, 600))
```

```python
# 1. Blurring (Smoothing)
blurred_image = cv.GaussianBlur(image, (5, 5), 0)
```

```python
# 2. Sharpening - Define a sharping kernel

sharpening_kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])

sharpened_image = cv.filter2D(image, -1, sharpening_kernel)
```

```python
# 3. Edge Detection (Using Canny)
edges = cv.Canny(image, 100, 200)
```

```python
# Display the images

cv.imshow('Original Image', image)
cv.imshow('Blurred Image', blurred_image)
cv.imshow('Sharpened Image', sharpened_image)
cv.imshow('Edge Detection (Canny)', edges)

cv.waitKey(0)
cv.destroyWindows()
```
