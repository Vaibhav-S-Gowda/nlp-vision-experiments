```python
import cv2
import matplotlib.pyplot as plt
```

### Read the image

```python
img = cv2.imread(r"C:\Users\vaibh\Downloads\serene-samurai-amidst-cherry-blossoms-moonlit-lake.jpg")
```

### Convert BGR to RGB for displaying with Matplotlib

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

### Create a copy for segmentation

```python
result = img.copy()
```

### Create a mask required by floodFill

```python
mask = cv2.UMat(1, 1, cv2.CV_8UC1)
```

### Choose a seed point inside the region

```python
seed_point = (250, 400)
```

### Fill the connected region with green

```python
cv2.floodFill(
    result, None, seed_point, (0, 255, 0),
    loDiff = (20, 20, 20),
    upDiff = (20, 20, 20)
)
```

### Display original and segmented images

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Region Growing using Flood Fill")
plt.axis("off")

plt.show()
```
