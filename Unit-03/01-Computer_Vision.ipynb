# Image Manipulation Technique
1) Cropping
2) Resizing
3) Rotating
4) Flipping

```python
import cv2
import numpy as np
```

```python
image = cv2.imread(r"C:\Users\vaibh\Downloads\japan-background-digital-art.jpg")
image
```

```python
if image is None:
    print("Error")
else:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
resized = cv2.resize(gray, (224, 224))
resized
```

```python
denoised = cv2.GaussianBlur(resized, (5,5), 0)
```

```python
normalized = denoised.astype(np.float32)/255.0
normalized
```

```python
enhanced_gray = cv2.equalizeHist(gray)
enhanced_gray
```

```python
cv2.imshow('Original image', image)
cv2.imshow('Grayscale image', gray)
cv2.imshow('Resized image', resized)
cv2.imshow('Denoised image', denoised)
cv2.imshow('Normalized image', normalized)
cv2.imshow('Contast Enhanced Grayscale image', enhanced_gray)

key = cv2.waitKey(0)
if key == 27:
    print("Escape key pressed, closing window...")
else:
    print(f"Key pressed: {key}")
cv2.destroyAllWindows()
```

```python
cropped_image = image[50: 300, 100: 400]
cropped_image
```

```python
rows, cols, _ = image.shape
center = (cols / 2, rows / 2)
rotation_matrix = cv2.getRotationMatrix2D(center, 45, 1)
rotated_image = cv2.warpAffine(image, rotation_matrix, (cols, rows))
```

```python
flipped_image = cv2.flip(image, 1)
```

```python
cv2.imshow('Original image', image)
cv2.imshow('Cropped image', cropped_image)
cv2.imshow('Resized image', resized)
cv2.imshow('Rotated image', rotated_image)
cv2.imshow('Flipped image', flipped_image)

key = cv2.waitKey(0)
if key == 27:
    print("Escape key pressed, closing window...")
else:
    print(f"Key pressed: {key}")
cv2.destroyAllWindows()
```
