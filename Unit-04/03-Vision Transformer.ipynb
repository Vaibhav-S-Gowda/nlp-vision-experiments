```python
import timm, torch
from PIL import Image
from torchvision import transforms

# from google.colab import files
```

```python
# img = Image.open(list(files.upload().keys())[0]).convert('RGB')

img = Image.open(r"C:\Users\vaibh\Downloads\serene-samurai-amidst-cherry-blossoms-moonlit-lake.jpg").convert("RGB")
```

```python
# model.timm.create_model("vit_base_patch16_224", pretrained=True).eval()

model = timm.create_model("vit_base_patch16_224", pretrained=True).eval()
```

# This line loads a ready-made Vision Transformer pretrained on ImageNet and prepares it for prediction

```python
prep = transforms.Compose([
    transforms.Resize(224), transforms.CenterCrop(224),
    transforms.ToTensor(), transforms.Normalize(
        (0.485, 0.456, 0.406), (0.229, 0.224, 0.225)
    )
])
```

### Tells PyTorch not to calculate gradients done during training gives the image to the Vision Transformer model and get the sores for all possible classes

```python
with torch.no_grad():
    logits = model(prep(img).unsqueeze(0))
pred = logits.argmax().item()
print("Predicted class index: ", pred)
```
