```python
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import requests
```

```python
# Loads a processor that helps prepare images for the model. Processor - BlipProcessor.from_pretrained ("Salesforce/blip-image-captioning-base")
# It is used to load a pre-trained model from hugging face model hub which generates text for image
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")
```

### Load your image 

```python
image_path = r"C:\Users\vaibh\Downloads\serene-samurai-amidst-cherry-blossoms-moonlit-lake.jpg"
image = Image.open(image_path)
```

### Preprocess the image

```python
inputs = processor(images = image, return_tensors = "pt")
```

### Generate a caption for the image

```python
# **input-unpack the dictionary
out = model.generate(**inputs)
```

### The generated token IDs are converted back into a text that describes the image.

```python
caption = processor.decode(out[0], skip_special_tokens = True)
image.show()
```

```python
print("Generated caption: ", caption)
```
