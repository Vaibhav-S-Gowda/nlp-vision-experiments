## Basic Tokenization

```python
import spacy
```

```python
### spacy language model - https://spacy.io/usage/models/
```

```python
# This will create a blank model
nlp = spacy.blank("en")
doc = nlp("In the Marvel Cinematic Universe, Earth's mightiest heroes including Iron Man, Captain America, Thor, Hulk, Black Widow, Spider-Man, Doctor Strange, and Black Panther—must join forces when Thanos, the powerful Mad Titan, discovers the six Infinity Stones and threatens to reshape the entire universe; meanwhile, Nick Fury secretly contacts new heroes such as Captain Marvel and Spider-Man, while the Avengers struggle with trust, sacrifice, friendship, and the difficult question: How can Earth's heroes possibly defeat an enemy who controls the power of the universe?")

for token in doc:
    print(token)
```

### Example

```python
doc = nlp('''"Tony's Avengers re-assembled in 2026—fast! Can Spider-Man's AI-powered suit handle 50+ enemies? #Marvel @StarkTech"''')
for token in doc:
    print(token)
```

```python
for token in doc:
    print(token.text, "| is_alpha:", token.is_alpha, "| is_digit:", token.is_digit, "| is_punct:", token.is_punct, "| is_symbol:", token.pos_ == "SYM")
```

# POS + Explanation

```python
doc = nlp("Elon flew to mars yesterday. He carried biriyani masala with him")

for token in doc:
    print(token, "|", token.pos_, "|",
         spacy.explain(token.pos_), "|",
         token.tag_, "|",
         spacy.explain(token.tag_))
```
