```python
# Load spaCy's English model
nlp = spacy.load("en_core_web_sm")
```

```python
# Example text
text = "Natural Language Processing with spaCy is powerful and intutive."

# Process the text using spaCy
doc = nlp(text)
```

```python
# 1. Tokenization
tokens = [token.text for token in doc]
print("Tokens: ", tokens)
```

```python
# 2. POS Tagging
pos_tags = [(token.text, token.pos_) for token in doc]
print("Part-of-speech Tags: ", pos_tags)
```

```python
# Print each word with its part of speech and its explanation
for token in doc:
    print(token, "|", token.pos_, "|", spacy.explain(token.pos_))

# spacy.explain() is a built-in function in the spaCy library
# returns a human-readable explanation
```

```python
count  = doc.count_by(spacy.attrs.POS)
count
```

```python
# token.text -> word
# token.pos_ -> human-readable POS such as NOUN, VERB, ADJ
# token.tag_ -> detailed POS tag such as NN, VB, JJ
# token.pos -> numeric POS ID
```

```python
for token in doc:
    print(token.text, token.pos_, token.tag_, token.pos)
```

```python
# 3. Named Entity Recognition (NER)
# Example text
text = "Dr. A. P. J. Abdul Kalam worked at ISRO in India."

# Process the text using spaCy
doc = nlp(text)

# Named Entity Recognition
print("Named Entities: ")

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

```python
# 4. Stemming
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stems  = [stemmer.stem(token.text) for token in doc]
print("Stemming: ", stems)
```

```python
# 5. Lemmatization
lemmas = [token.lemma_ for token in doc]
print("Lemmatization: ", lemmas)
```

```python
# 6. Stop Words
# Removing stop words
filtered_tokens = [token.text for token in doc if not token.is_stop]
print("Without Stop Words: ", filtered_tokens)
```

```python
# 7. Bag of Words (BoW)
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform([text])
print("Bag of Words (BoW): ")
print(vectorizer.get_feature_names_out())
print(bow_matrix.toarray())
```

```python
# 8. N-grams (e.g., bi-grams)
bigram_vectorizer = CountVectorizer(ngram_range = (2, 2))
bigram_matrix = bigram_vectorizer.fit_transform([text])
print("Bi-grams: ")
print(bigram_vectorizer.get_feature_names_out())
print(bigram_matrix.toarray())
```
