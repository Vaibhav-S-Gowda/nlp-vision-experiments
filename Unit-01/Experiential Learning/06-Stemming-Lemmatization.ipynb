# Stemming
### PorterStemmer is a class
### which is imported from nltk.stem module

```python
from nltk.stem import PorterStemmer  # Snowball Stemmer
stemmer = PorterStemmer()
```

```python
words = ['eating', 'eats', 'ate', 'adjustable', 'rafting', 'ability', 'meeting']

for word in words:
    print(word, "|", stemmer.stem(word))
```

# Lemmatization

### Load the spaCy English Language model

```python
import spacy

nlp = spacy.load("en_core_web_sm")

# Input text to process
doc = nlp("eaating eats eat ate adjustable rafting ability meeting")

# Iterate through each token in the processed text
for token in doc:
    print(token, "|", token.lemma_)
```
