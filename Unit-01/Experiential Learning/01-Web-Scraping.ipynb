### Web scraping is the process of extracting data from different sources instead of copying

## Web Scraping JSON data

```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://wikimediafoundation.org/jobs/'
```

```python
# Add headers to mimic a real browser
# headers = {
#     'User-Agent': 'Mozilla/5.0 (Windows NT 10.0;)'
# }
```

```python
# sends a request to the server to get the webpage
response = requests.get(url)

# Contains all the HTML content of the page
print(response)
```

```python
# Parse HTML
soup = BeautifulSoup(response.text, 'html.parser')

# Extract paragrraph text
text = ''
for paragraph in soup.find_all('p'):
    text += paragraph.get_text() + '\n'

print(text[:1000])
```
