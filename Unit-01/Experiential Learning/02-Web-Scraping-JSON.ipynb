## Web Scrapping JSON data

```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://saurav.tech/NewsAPI/top-headlines/category/technology/in.json'
```

```python
# sends a request to the server to get the webpage
response = requests.get(url)

# Contains all the HTML content of the page
print(response)
```

```python
data = response.json()
```

```python
for book in data['articles']:
    print(book['title'])
```
