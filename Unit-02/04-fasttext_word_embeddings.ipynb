```python
from gensim.models import FastText
```

```python
sentences = [
    ["i", "love", "nlp"],
    ["nlp", "is", "fun"],
    ["word", "embeddings", "capture", "meaning"],
    ["fasttext", "uses", "subword", "information"]
]
```

## Train FastText model

```python
model = FastText(sentences, vector_size = 50, window = 3,
                min_count = 1, epochs = 10)
```

## Check vector for a word

```python
print("Vector for 'nlp': ")
print(model.wv['nlp'])
```
