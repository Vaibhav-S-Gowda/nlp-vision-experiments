```python
import gensim.downloader as api
model = api.load('glove-wiki-gigaword-50')
```

## Check vector for a word

```python
print("Vector for 'King' : ")
print(model['king'])
```

## Find similar words

```python
print("Top 5 similar words to 'King': ")
print(model.most_similar('king', topn=5))
```

## Find similarity between two words

```python
print("Similarity between 'King' and 'Queen': ", model.similarity('king', 'queen'))
print("Similarity between 'King'' and 'Apple': ", model.similarity('king', 'apple'))
```
