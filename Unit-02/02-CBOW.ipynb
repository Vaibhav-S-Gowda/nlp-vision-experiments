```python
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess
```

## Take input sentence from user

```python
text = input("Enter a sentence: ")
```

## Preprocess sentence: convert to list of words

```python
sentences = [simple_preprocess(text)]
```

## Train CBOW model (sg=0)

```python
cbow_model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, sg=0)
```

## Train skip-Gram model (sg=1)

```python
skip_model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, sg=1)
```

## Ask user for a word to check similarity

```python
word = input("Enter a word to check similarity: ")
```

## Show similar words from CBOW

```python
print("\nCBOW similar to '{}': ".format(word))
if word in cbow_model.wv:
    print(cbow_model.wv.most_similar(word))
else: 
    print("Word not in corpus!")
```
