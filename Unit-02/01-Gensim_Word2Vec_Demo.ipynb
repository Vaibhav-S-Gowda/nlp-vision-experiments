```python
from  gensim.models import Word2Vec
```

```python
sentences = [
    ["the", "cat", "sits", "on", "the", "mat"],
    ["the", "dog", "plays", "in", "the", "garden"],
    ["the", "cat", "and", "dog", "are", "friends"]
]
```

```python
cbow_model = Word2Vec(sentences, vector_size=20, window=2, min_count=1, sg=0)
similar_words = cbow_model.wv.most_similar("cat")
print("CBOW - Words similar to 'cat' : ")
for word, score in similar_words:
    print(word, ":", round(score, 4))
```

```python
skip_model = Word2Vec(sentences, vector_size=20, window=2, min_count=1, sg=1)
similar_words = skip_model.wv.most_similar('cat')
print("Skip-Gram similar to 'cat': ")
for word, score in similar_words:
    print(word, ":", round(score, 4))
```
